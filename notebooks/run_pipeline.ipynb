```python
config = XFuseConfig.from_yaml('configs/example_diad.yaml')
```

## Stage 0 — validate and prepare data (CPU, fast)

Inspect `outputs/<name>/data/data_overview.png` before proceeding.

```python
run_data(config)
```

## Stage 1 — extract features (GPU required)

Inspect PCA grid in `outputs/<name>/features/` before setting threshold.

```python
run_features(config)
```

## Stage 2 — threshold to binary masks

Set `threshold` after inspecting the PCA component 0 range.

```python
threshold = 0.2
run_segment(config, threshold)
```

## Stage 3 — SAM2 refinement

```python
run_refine(config)
```
